# WATER JUG PROBLEM

- You are given two jugs, 4-ltr and 3-ltr.
- Neither jug has any measuring markings on it.
- A pump which has unlimited water which you can choose to fill the jug.
- How can you get exactly 2 ltr of water in a 4 tr jug?
- We will represent a state of the problem as a tuple(x,y).
- x = amount of water in the 4 ltr jug
- y = amount of water in the 3 ltr jug
1) Initial state : (0,0); 0<=x<=4, and 0<=x<=3.
2) Goal state : (2,n) for any value of n.


# Production Rules
|Sr. | Current State | Next State | Description|
|------|--------|---------|--------|
|1 | (x,y) if x<4| (4,y) | Fill the 4-ltr jug|
|2 | (x,y) if y<3| (x,3) | Fill the 3-ltr jug|
|3| (x,y) if x>0| (x-d,y) | Pour some water out of the 4-ltr jug|
|4| (x,y) if y>0| (x,y-d) | Pour some water out of the 3-ltr jug|
|5| (x,y) if x>0| (0,y) | Empty the 4-ltr jug on the ground|
|6| (x,y) if y>0| (x,0) | Empty the 3-ltr jug on the ground|
|7| (x,y) i fx+y>=4 & y>0| (4,y-(4-x)) | Pour water from the 3-ltr jug into the 4-ltr jug until the 4-ltr jug is full|
|8| (x,y) if x+y>=3 & x>0| (x-(3-y),3) | Pour water from the 4-ltr jug into the 3-ltr jug until the 3-ltr jug is full|
|9| (x,y) if x+y<=4 & y>0| (x+y,0) | Pour all the water from the 3-ltr jug into the  4-ltr jug|
|10| (x,y) if x+y<=3 & x>0| (0,x+y) | Pour all the water from the 4-ltr jug into the  3-ltr jug|
|11| (0,2) | (2,0) | Pour the 2-ltr from the 3-ltr jug into the  4-ltr jug|
|12| (2,y) | (0,y) | Empty the 2-ltr in the 4-ltr on the ground|

# 8 PUZZLE PROBLEM

The 8 puzzle consists of eight numbered, movable tile set in a 3x3 frame. One cell of the frame is always empty thus making it possible to move an adjacent numbered tile into the empty cell.Such a puzzle is illustrated in following diagram.

# Rules of the Puzzle

- Only one tile can be moved at a time.

- A tile can move up, down, left, or right.

- A tile can move only into the blank space.

- The aim is to reach the goal state arrangement.

Initial state
|2|8|3|                                            
|-|-|-|                                           
|1|6|4|        
|7|_|5|                           

Goal state
|1|2|3|                                            
|-|-|-|                                           
|8|_|4|        
|7|6|5|

# Step-by-Step Solution
Step 1 – Move 6 Down
|2|8|3|
|-|-|-| 
|1|_|4|        
|7|6|5|

Tile 6 moves down into the blank space.

Step 2 – Move 1 Right
|2|8|3|
|-|-|-| 
|_|1|4|        
|7|6|5|

Tile 1 moves right into the blank space.

Step 3 – Move 2 Down
|_|8|3|
|-|-|-| 
|2|1|4|        
|7|6|5|

Tile 2 moves down.

Step 4 – Move 8 Left
|8|_|3|
|-|-|-| 
|2|1|4|        
|7|6|5|

Tile 8 moves left into the blank space.

Step 5 – Move 1 Up
|8|1|3|
|-|-|-| 
|2|_|4|        
|7|6|5|

Tile 1 moves up.

Step 6 – Move 2 Right
|8|1|3|
|-|-|-| 
|_|2|4|        
|7|6|5|

Tile 2 moves right.

Step 7 – Move 8 Down
|_|1|3|
|-|-|-| 
|8|2|4|        
|7|6|5|

Tile 8 moves down.

Step 8 – Move 1 Left
|1|_|3|
|-|-|-| 
|8|2|4|        
|7|6|5|

Tile 1 moves left.

Step 9 – Move 2 Up
|1|2|3|
|-|-|-| 
|8|_|4|        
|7|6|5|

Now the puzzle reaches the goal state.

# Final Goal State
|1|2|3|
|-|-|-| 
|8|_|4|        
|7|6|5|

The 8 Puzzle Problem demonstrates how search algorithms in AI find a sequence of moves from the initial state to the goal state.
By moving tiles step by step using the blank space, the system reaches the desired arrangement.

```python
# Implement PCA on a given matrix.

# step 1 collect data
X= np.array([[2, 3, 4], [3, 5, 6], [5, 8, 9], [6, 9, 11], [7, 10, 13]])

# step 2
# compute mean value for each column
mean_x = np.mean(X,axis= 0)
print(mean_x)
# Step 3: center the data
x_centered = X-mean_x
print(x_centered)
# Stpe 4: computer the covariance of the matrix
cov_matrix = np.cov(x_centered,rowvar=False)
print(cov_matrix)
# Step 5: computer eignenvalues and eigenvectore
eigenvalues,eigenvecotes = np.linalg.eig(cov_matrix)
print(eigenvalues)
print(eigenvecotes)
# Stpe 6: select the principle component(soriting by largest eigenvalues)
sorted_indices = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[sorted_indices]
eigenvecotes = eigenvecotes[:,sorted_indices]
# Step 7: Transform the dataset 
# 3 to 1 (project onto first principle component)
X_transformerd_1 = x_centered @ eigenvecotes[:,0]
print("\nTransformed Data (3 to 1):\n", X_transformerd_1)

# 3 to 2 (project onto first two princple components0)
X_transformerd_2 = x_centered @ eigenvecotes[:,:2]
print("\nTransformed Data (3 to 2):\n",X_transformerd_2)

# output result
print('Transformed Data (3 to 1):',X_transformerd_1)
print('Transformed Data (3 to 2):\n',X_transformerd_2)

```

```python
# KMedoids clustering on iris dataset using pyclustering.
# pip install pyclustering
import numpy as np
import pandas as pd
from pyclustering.cluster.kmedoids import kmedoids
from pyclustering.utils import distance_metric, type_metric
from sklearn import datasets
from sklearn.preprocessing import StandardScaler

# Load the wine dataset
wine = datasets.load_wine()

# convert the data into a pandas DataFeame for better readability 
winde_df = pd.DataFrame(wine.data,columns=wine.feature_names)

# standardize the dataset/
scaler = StandardScaler()
X_scaled = scaler.fit_transform(wine.data)

# Intial medoids(can be randomly chosen or specified)
# Here we choose the first three data points as inital medoids
initial_medoids = [0,1,2]

# # Randomly select initial medoids 
# # choose 3 random indices from the dataset
import random
initial_medoids = random.sample(range(len(X_scaled)), 3)

# create the K-Medoids isinstance
metric = distance_metric(type_metric.EUCLIDEAN)
kmedoids_instance = kmedoids(X_scaled, initial_medoids, metric=metric)

# perform the clustering
kmedoids_instance.process()

# Get the resulting cluseter and medoids 
clusters = kmedoids_instance.get_clusters()
medoid = kmedoids_instance.get_medoids()
print("Clusters:", clusters)
print("Medoids:", medoid)

# plot the results
plt.figure(figsize=(8,6))
for cluster in clusters:
    plt.scatter(X_scaled[cluster,0],X_scaled[cluster,1],label=f'cluster {clusters.index(cluster)}')
# Mark the medoids in red
for medoids in medoid:
    plt.scatter(X_scaled[medoids,0],X_scaled[medoids,1], s=200,c='red',marker='X',label='Medoid')
plt.title('K-Medoids clustering (Wine Dataset) with Pyclustering')
plt.legend()
plt.show()
```

```python
# Non- Linear SVM(KERNEL SVM)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.svm import SVC

X, y = datasets.make_circles(n_samples=200, factor=0.3, noise= 0.1, random_state= 42)

plt.figure()
plt.scatter(X[:,0], X[:,1], c=y)
plt.title("Before Applying Kernel SVM(Non-Linear Data)")
plt.xlabel("Features 1")
plt.ylabel("Features 2")
plt.show
```

```python
# Training a Kernel SVM (RBF Kernel)
from sklearn.svm import SVC

# Initialize the SVC model with an RBF kernel
kernel_svm = SVC(kernel='rbf', random_state=42)

# Train the model
kernel_svm.fit(X, y)

print("Kernel SVM (RBF) trained successfully!")

def plot_decision_boundary(model, X, y, title):
    plt.figure(figsize=(8, 6))
    # Create a mesh to plot in
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02),
                         np.arange(y_min, y_max, 0.02))

    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.contourf(xx, yy, Z, alpha=0.8, cmap=plt.cm.coolwarm)
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.coolwarm, s=50, edgecolors='k')
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.title(title)
    plt.show()

# Re-generate X and y for the Kernel SVM plot, as they were overwritten later
X_svm, y_svm = datasets.make_circles(n_samples=200, factor=0.3, noise= 0.1, random_state= 42)

plot_decision_boundary(kernel_svm, X_svm, y_svm, "Kernel SVM (RBF) Decision Boundary")
```
